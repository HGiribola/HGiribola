```python
FILE_PATH = 'Global_Cybersecurity_Threats_2015-2024 (1).csv'
df = pd.read_csv(FILE_PATH)

print("Dimensões do dataset:", df.shape)
print("\nColunas:")
print(df.columns)

print("\nPrimeiras linhas:")
print(df.head())
```

```python
target_col = "Financial Loss (in Million $)"

print("\nDescrição estatística da variável alvo:")
print(df[target_col].describe())

print("\nVerificando valores ausentes por coluna:")
print(df.isna().sum())
```

```python
y = df[target_col]
```

```python
X = df.drop(columns=[target_col])
```

```python
X_enc = pd.get_dummies(X, drop_first=True)
```

```python
print("\nShape de X após get_dummies:", X_enc.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_enc,
    y,
    test_size=0.2,
    random_state=42
)

print("\nTamanho treino:", X_train.shape, " | Tamanho teste:", X_test.shape)
```

```python
max_depth = 4

dt_model = DecisionTreeRegressor(
    max_depth=max_depth,
    random_state=42
)

dt_model.fit(X_train, y_train)

y_pred_dt = dt_model.predict(X_test)

rmse_dt = math.sqrt(mean_squared_error(y_test, y_pred_dt))
r2_dt = r2_score(y_test, y_pred_dt)

print("\n=== Árvore de Decisão (Regressão) ===")
print(f"Profundidade máxima: {max_depth}")
print(f"RMSE (teste): {rmse_dt:.4f}")
print(f"R² (teste):   {r2_dt:.4f}")

# Importância das variáveis na Árvore
feat_importances_dt = pd.DataFrame({
    "feature": X_enc.columns,
    "importance": dt_model.feature_importances_
}).sort_values(by="importance", ascending=False)

print("\nTop 15 importâncias de features – Árvore de Decisão:")
print(feat_importances_dt.head(15))
```

```python
rf_model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

rmse_rf = math.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("\n=== Random Forest (Regressão) ===")
print(f"n_estimators: 300")
print(f"RMSE (teste): {rmse_rf:.4f}")
print(f"R² (teste):   {r2_rf:.4f}")


feat_importances_rf = pd.DataFrame({
    "feature": X_enc.columns,
    "importance": rf_model.feature_importances_
}).sort_values(by="importance", ascending=False)

print("\nTop 15 importâncias de features – Random Forest:")
print(feat_importances_rf.head(15))
```

```python
print("\n=== Comparação Geral ===")
print("Árvore de Decisão -> RMSE: {:.4f} | R²: {:.4f}".format(rmse_dt, r2_dt))
print("Random Forest     -> RMSE: {:.4f} | R²: {:.4f}".format(rmse_rf, r2_rf))
```
